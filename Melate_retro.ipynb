```python
# -*- coding: utf-8 -*-
"""Sistema Predictor Melate Retro - Versión Estable"""

# Configuración inicial post-reinicio
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import layers, models

# Carga de datos segura
from google.colab import files

def cargar_datos():
    uploaded = files.upload()
    filename = list(uploaded.keys())[0]
    df = pd.read_csv(filename)

    # Verificación de columnas
    columnas_requeridas = ['N1', 'N2', 'N3', 'N4', 'N5', 'N6', 'Adicional']
    if not all(col in df.columns for col in columnas_requeridas):
        raise ValueError("El CSV no tiene las columnas requeridas")
    return df[columnas_requeridas]

try:
    datos = cargar_datos()
except Exception as e:
    print(f"Error: {e}. Usando datos de ejemplo.")
    datos = pd.DataFrame({
        'N1': [4, 12, 8], 'N2': [10, 15, 22],
        'N3': [15, 18, 25], 'N4': [20, 28, 30],
        'N5': [25, 35, 36], 'N6': [30, 38, 40],
        'Adicional': [7, 9, 11]
    })

# Preprocesamiento
scaler = tf.keras.layers.Normalization(axis=-1)
scaler.adapt(datos.values)

# Modelo mínimo viable
modelo = models.Sequential([
    layers.Dense(32, activation='relu', input_shape=(7,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(7)  # 6 números + adicional
])

modelo.compile(
    optimizer='adam',
    loss='mse'
)

# Entrenamiento
X = datos.iloc[:-1]
y = datos.iloc[1:]  # Predicción del siguiente sorteo

history = modelo.fit(
    scaler(X),
    scaler(y),
    epochs=50,
    batch_size=4,
    verbose=1
)

# Predicción
ultimo_sorteo = datos.iloc[-1:].values
prediccion = scaler(modelo.predict(scaler(ultimo_sorteo)))
print("\n🔮 Predicción:", np.round(prediccion).astype(int)[0])
```

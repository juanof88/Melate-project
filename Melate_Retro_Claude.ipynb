```python
# -*- coding: utf-8 -*-
"""
SISTEMA PREDICTOR MELATE RETRO - ARQUITECTURA LSTM COMPLETA
Dataset: 1,615 sorteos históricos
Features: 39 variables derivadas + embeddings
"""

# ============================================================================
# 1. INSTALACIÓN DE DEPENDENCIAS
# ============================================================================

!pip install -q pandas numpy tensorflow scikit-learn matplotlib seaborn

# ============================================================================
# 2. IMPORTS
# ============================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from tensorflow.keras.optimizers import Adam
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

print("✅ Librerías importadas")
print(f"TensorFlow version: {tf.__version__}")

# ============================================================================
# 3. CARGA DE DATOS
# ============================================================================

from google.colab import files

print("\n📁 Sube tu archivo CSV del Melate Retro:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]

df = pd.read_csv(filename)

# Mostrar info del dataset
print(f"\n✅ Datos cargados: {len(df)} sorteos")
print(f"Columnas: {list(df.columns)}")
print(f"\nPrimeros 5 sorteos:")
print(df.head())

# Verificar estructura de columnas
print("\n🔍 Detectando estructura del dataset...")
if 'F1' in df.columns:
    print("   ✅ Formato detectado: F1-F6 (principales), F7 (adicional)")
    NUMEROS_COLS = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6']
    ADICIONAL_COL = 'F7'
elif 'N1' in df.columns:
    print("   ✅ Formato detectado: N1-N6 (principales), Adicional")
    NUMEROS_COLS = ['N1', 'N2', 'N3', 'N4', 'N5', 'N6']
    ADICIONAL_COL = 'Adicional'
else:
    raise ValueError("❌ No se detectaron las columnas esperadas (F1-F6 o N1-N6)")

print(f"   Números principales: {NUMEROS_COLS}")
print(f"   Número adicional: {ADICIONAL_COL}")

# ============================================================================
# 4. GENERACIÓN DE FEATURES
# ============================================================================

print("\n🔧 Generando features avanzados...")

# Features básicos
df['suma'] = df[NUMEROS_COLS].sum(axis=1)
df['media'] = df[NUMEROS_COLS].mean(axis=1)
df['std'] = df[NUMEROS_COLS].std(axis=1)
df['min'] = df[NUMEROS_COLS].min(axis=1)
df['max'] = df[NUMEROS_COLS].max(axis=1)
df['rango'] = df['max'] - df['min']

# Pares/Impares
def contar_pares(row):
    nums = row[NUMEROS_COLS].values
    return sum(n % 2 == 0 for n in nums)

df['pares'] = df.apply(contar_pares, axis=1)
df['impares'] = 6 - df['pares']

# Rangos (bajo/medio/alto)
def contar_rangos(row):
    nums = row[NUMEROS_COLS].values
    bajo = sum(n <= 13 for n in nums)
    medio = sum(13 < n <= 26 for n in nums)
    alto = sum(n > 26 for n in nums)
    return bajo, medio, alto

rangos = df.apply(contar_rangos, axis=1, result_type='expand')
df['bajo'] = rangos[0]
df['medio'] = rangos[1]
df['alto'] = rangos[2]

# Consecutivos
def contar_consecutivos(row):
    nums = sorted(row[NUMEROS_COLS].values)
    return sum(nums[i+1] - nums[i] == 1 for i in range(5))

df['consecutivos'] = df.apply(contar_consecutivos, axis=1)

# Repeticiones del sorteo anterior
repeticiones = [0]
for i in range(1, len(df)):
    anterior = set(df.iloc[i-1][NUMEROS_COLS].values)
    actual = set(df.iloc[i][NUMEROS_COLS].values)
    repeticiones.append(len(anterior & actual))

df['rep_anterior'] = repeticiones

# Top 10 números más frecuentes
numeros_principales = df[NUMEROS_COLS].values.flatten()
freq_counter = Counter(numeros_principales)
numeros_top = [num for num, _ in freq_counter.most_common(10)]

print(f"Top 10 números: {numeros_top}")

# Frecuencias acumuladas
for num in numeros_top:
    frecuencias = []
    contador = 0
    for idx, row in df.iterrows():
        if num in row[NUMEROS_COLS].values:
            contador += 1
        frecuencias.append(contador)
    df[f'freq_{num}'] = frecuencias

# Gaps
for num in numeros_top:
    gaps = []
    ultimo_idx = -1
    for idx, row in df.iterrows():
        if num in row[NUMEROS_COLS].values:
            ultimo_idx = idx
        if ultimo_idx == -1:
            gaps.append(idx + 1)
        else:
            gaps.append(idx - ultimo_idx)
    df[f'gap_{num}'] = gaps

# Medias móviles
df['suma_ma_5'] = df['suma'].rolling(window=5, min_periods=1).mean()
df['suma_ma_10'] = df['suma'].rolling(window=10, min_periods=1).mean()
df['suma_ma_20'] = df['suma'].rolling(window=20, min_periods=1).mean()

# Volatilidad
df['suma_std_5'] = df['suma'].rolling(window=5, min_periods=1).std().fillna(0)
df['suma_std_10'] = df['suma'].rolling(window=10, min_periods=1).std().fillna(0)

print(f"✅ Features generados: {len(df.columns)} columnas totales")

# ============================================================================
# 5. PREPARACIÓN DE DATOS PARA LSTM
# ============================================================================

print("\n🔄 Preparando datos para LSTM...")

# Seleccionar features para el modelo
# Excluir columnas originales del CSV y features derivados específicos
exclude_cols = NUMEROS_COLS + [ADICIONAL_COL]
if 'NPRODUCTO' in df.columns:
    exclude_cols.extend(['NPRODUCTO', 'CONCURSO', 'BOLSA', 'FECHA'])

feature_cols = [col for col in df.columns if col not in exclude_cols]
target_cols = NUMEROS_COLS

print(f"Features de entrada: {len(feature_cols)}")
print(f"Targets: {target_cols}")

# Normalizar features
scaler_features = StandardScaler()
df_scaled = df.copy()
df_scaled[feature_cols] = scaler_features.fit_transform(df[feature_cols])

# Normalizar targets (números 1-39)
scaler_targets = MinMaxScaler(feature_range=(0, 1))
df_scaled[target_cols] = scaler_targets.fit_transform(df[target_cols])

# Crear secuencias para LSTM
def create_sequences(data, features, targets, seq_length=10):
    X, y = [], []
    for i in range(seq_length, len(data)):
        X.append(data[features].iloc[i-seq_length:i].values)
        y.append(data[targets].iloc[i].values)
    return np.array(X), np.array(y)

SEQ_LENGTH = 10
X, y = create_sequences(df_scaled, feature_cols, target_cols, SEQ_LENGTH)

print(f"\n✅ Secuencias creadas:")
print(f"   X shape: {X.shape} (samples, timesteps, features)")
print(f"   y shape: {y.shape} (samples, números)")

# Split temporal (80% train, 20% test)
split_idx = int(len(X) * 0.8)
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

print(f"\n📊 Split de datos:")
print(f"   Train: {len(X_train)} secuencias")
print(f"   Test: {len(X_test)} secuencias")

# ============================================================================
# 6. CONSTRUCCIÓN DEL MODELO LSTM
# ============================================================================

print("\n🏗️ Construyendo modelo LSTM...")

def build_lstm_model(input_shape, output_dim):
    model = models.Sequential([
        # Primera capa LSTM con return_sequences
        layers.LSTM(128, return_sequences=True, input_shape=input_shape),
        layers.Dropout(0.3),

        # Segunda capa LSTM
        layers.LSTM(64, return_sequences=False),
        layers.Dropout(0.3),

        # Capas densas
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.2),
        layers.Dense(64, activation='relu'),
        layers.Dense(output_dim, activation='sigmoid')  # Sigmoid para valores 0-1
    ])

    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='mse',
        metrics=['mae']
    )

    return model

model = build_lstm_model(
    input_shape=(X_train.shape[1], X_train.shape[2]),
    output_dim=y_train.shape[1]
)

print("\n📐 Arquitectura del modelo:")
model.summary()

# ============================================================================
# 7. CALLBACKS Y ENTRENAMIENTO
# ============================================================================

print("\n🎯 Entrenando modelo...")

# Callbacks
early_stop = callbacks.EarlyStopping(
    monitor='val_loss',
    patience=15,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    verbose=1
)

# Entrenamiento
history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=100,
    batch_size=32,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

print("\n✅ Entrenamiento completado")

# ============================================================================
# 8. EVALUACIÓN
# ============================================================================

print("\n📊 Evaluando modelo...")

# Predicciones
y_pred_scaled = model.predict(X_test)

# Desnormalizar predicciones
y_pred = scaler_targets.inverse_transform(y_pred_scaled)
y_test_original = scaler_targets.inverse_transform(y_test)

# Redondear a enteros
y_pred_rounded = np.round(y_pred).astype(int)
y_pred_rounded = np.clip(y_pred_rounded, 1, 39)  # Asegurar rango 1-39

# Métricas
mae = mean_absolute_error(y_test_original, y_pred)
rmse = np.sqrt(mean_squared_error(y_test_original, y_pred))

print(f"\n📈 Métricas de evaluación:")
print(f"   MAE: {mae:.2f}")
print(f"   RMSE: {rmse:.2f}")

# Accuracy por posición (número exacto)
accuracies = []
for i in range(6):
    accuracy = np.mean(y_pred_rounded[:, i] == y_test_original[:, i])
    accuracies.append(accuracy)
    print(f"   Accuracy N{i+1}: {accuracy*100:.2f}%")

print(f"   Accuracy promedio: {np.mean(accuracies)*100:.2f}%")

# Accuracy de "al menos 1 acierto"
at_least_one = 0
for i in range(len(y_test_original)):
    if len(set(y_pred_rounded[i]) & set(y_test_original[i].astype(int))) > 0:
        at_least_one += 1

print(f"   Al menos 1 acierto: {at_least_one/len(y_test_original)*100:.2f}%")

# ============================================================================
# 9. VISUALIZACIÓN
# ============================================================================

print("\n📉 Generando visualizaciones...")

# Gráfica de pérdida
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Pérdida durante entrenamiento')
plt.xlabel('Época')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['mae'], label='Train MAE')
plt.plot(history.history['val_mae'], label='Val MAE')
plt.title('MAE durante entrenamiento')
plt.xlabel('Época')
plt.ylabel('MAE')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Comparación predicciones vs reales (últimos 10 sorteos)
print("\n🔮 Últimos 10 sorteos - Predicción vs Real:")
print("="*80)

for i in range(max(0, len(y_test_original)-10), len(y_test_original)):
    real = sorted(y_test_original[i].astype(int))
    pred = sorted(y_pred_rounded[i])
    aciertos = set(real) & set(pred)

    print(f"\nSorteo {i+1}:")
    print(f"  Real:       {real}")
    print(f"  Predicción: {pred}")
    print(f"  Aciertos:   {len(aciertos)} números {list(aciertos) if aciertos else ''}")

# ============================================================================
# 10. PREDICCIÓN DEL PRÓXIMO SORTEO
# ============================================================================

print("\n" + "="*80)
print("🎲 PREDICCIÓN DEL PRÓXIMO SORTEO")
print("="*80)

# Tomar las últimas SEQ_LENGTH observaciones
last_sequence = df_scaled[feature_cols].iloc[-SEQ_LENGTH:].values
last_sequence = last_sequence.reshape(1, SEQ_LENGTH, -1)

# Predecir
next_pred_scaled = model.predict(last_sequence, verbose=0)
next_pred = scaler_targets.inverse_transform(next_pred_scaled)
next_pred_rounded = np.round(next_pred).astype(int)[0]
next_pred_rounded = np.clip(next_pred_rounded, 1, 39)

# Asegurar que no haya duplicados
next_pred_unique = []
for num in next_pred_rounded:
    if num not in next_pred_unique:
        next_pred_unique.append(num)

# Si hay menos de 6, completar con números aleatorios no usados
while len(next_pred_unique) < 6:
    random_num = np.random.randint(1, 40)
    if random_num not in next_pred_unique:
        next_pred_unique.append(random_num)

next_pred_final = sorted(next_pred_unique[:6])

print(f"\n🔮 Predicción del próximo sorteo:")
print(f"   Números: {next_pred_final}")
print(f"   Suma: {sum(next_pred_final)}")
print(f"   Pares: {sum(1 for n in next_pred_final if n % 2 == 0)}")
print(f"   Impares: {sum(1 for n in next_pred_final if n % 2 != 0)}")

# Contexto estadístico
ultimo_sorteo = df[NUMEROS_COLS].iloc[-1].values
print(f"\n📊 Último sorteo real: {sorted(ultimo_sorteo)}")
print(f"   Suma: {sum(ultimo_sorteo):.0f}")

print("\n" + "="*80)
print("⚠️  DISCLAIMER:")
print("Este modelo es un ejercicio académico de Deep Learning.")
print("Las loterías son eventos aleatorios. No hay garantía de aciertos.")
print("="*80)

# ============================================================================
# 11. GUARDAR MODELO
# ============================================================================

print("\n💾 ¿Deseas guardar el modelo?")
print("   Descomenta las siguientes líneas para guardar:")
print()
print("# model.save('melate_lstm_model.h5')")
print("# import pickle")
print("# with open('scalers.pkl', 'wb') as f:")
print("#     pickle.dump({'features': scaler_features, 'targets': scaler_targets}, f)")
print("# print('✅ Modelo guardado')")

print("\n🎉 ¡Análisis completado!")
```
